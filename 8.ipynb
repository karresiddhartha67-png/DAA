```python

from collections import defaultdict

graph = defaultdict(list)

def add_edge(u, v):
    graph[u].append(v)
    graph[v].append(u)

def dfs(start):
    visited = set()
    stack = [start]
    result = []

    while stack:
        node = stack.pop()
        if node not in visited:
            visited.add(node)
            result.append(node)
            for neighbor in reversed(graph[node]):
                if neighbor not in visited:
                    stack.append(neighbor)
    return result


# Example
add_edge(0, 1)
add_edge(0, 2)
add_edge(1, 3)
add_edge(1, 4)
add_edge(2, 5)
add_edge(2, 6)

print("DFS:", dfs(0))
```

```python
from collections import defaultdict, deque

graph = defaultdict(list)

def add_edge(u, v):
    graph[u].append(v)
    graph[v].append(u)

def bfs(start):
    visited = set()
    queue = deque([start])
    visited.add(start)
    result = []

    while queue:
        node = queue.popleft()
        result.append(node)
        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)
    return result

# Example
add_edge(0, 1)
add_edge(0, 2)
add_edge(1, 3)
add_edge(1, 4)
add_edge(2, 5)
add_edge(2, 6)

print("BFS:", bfs(0))
```
